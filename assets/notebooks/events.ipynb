```
colors = ColorData[97, "ColorList"];
```

```
MakeEvents[dist_, add_] := Select[
    Accumulate[RandomVariate[dist, 1000]+add],
    # < 100 &
]
```

```
MakePlot[dist_, add_] := TimelinePlot[
    MakeEvents[dist, add],
    PlotRange->{0,100},
    PlotRangePadding -> None,
    ImageSize -> 1024,
    PlotMarkers -> "|",
    AxesStyle -> Large,
    FrameTicks -> {{None, None}, {Table[{t, Style[IntegerString[t], Large]}, {t, 0, max, 10}], None}}

]
```

```
exp = Show[MakePlot[ExponentialDistribution[1/3], 0], PlotLabel->Style["Exponential(1/3)", Large]]
```

```
Mean@ExponentialDistribution[1/3]
```

```
geom = Show[MakePlot[GeometricDistribution[1/3], 1], PlotLabel->Style["Geometric(1/3)", Large]]
```

```
eventTimes = MakeEvents[ExponentialDistribution[1/3], 0];
```

```
counts = Show[
    DateHistogram[eventTimes, {0, 100, 10}, ChartStyle->colors[[1]]],
    ListPlot[{#,1}& /@ eventTimes, PlotMarkers->Style["|", Large, colors[[2]]]],
    ImageSize -> 1024,
    PlotRangePadding -> None,
    Frame->True,
    FrameTicks -> {{Table[{y,Style[IntegerString[y], Large]}, {y, 0, 8}], None}, {Table[{t, Style[IntegerString[t], Large]}, {t, 0, 100, 10}], None}},
    FramePadding -> None,
    PlotLabel->Style["Counts of events with Exponential(1/3) waiting times for T=10", Large]

]
```

```
Export["../images/exp_events_count.png", counts, Background->None];
```
